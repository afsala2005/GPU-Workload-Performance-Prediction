```python
import numpy as np
import pandas as pd

from sklearn.model_selection import GroupKFold
from sklearn.base import clone
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

models = {
    'Tuned Random Forest (Log Target)': rf_tuned,
    'XGBoost (Log Target)': xgb_model
}

group_cv = GroupKFold(n_splits=5)

results = []

for name, model in models.items():

    fold_mae = []
    fold_rmse = []
    fold_r2 = []

    for train_idx, test_idx in group_cv.split(X, y_log, groups):

        X_train_fold = X.iloc[train_idx]
        X_test_fold = X.iloc[test_idx]

        y_train_fold = y_log.iloc[train_idx]
        y_test_fold = y.iloc[test_idx]

        # Create a fresh model for each fold
        current_model = clone(model)

        # Train using log-transformed target
        current_model.fit(X_train_fold, y_train_fold)

        # Predict in log scale
        predictions_log = current_model.predict(X_test_fold)

        # Convert predictions back to milliseconds
        predictions = np.expm1(predictions_log)

        # Calculate metrics on original scale
        fold_mae.append(
            mean_absolute_error(y_test_fold, predictions)
        )

        fold_rmse.append(
            np.sqrt(mean_squared_error(y_test_fold, predictions))
        )

        fold_r2.append(
            r2_score(y_test_fold, predictions)
        )

    results.append({
        'Model': name,
        'Mean MAE': np.mean(fold_mae),
        'Std MAE': np.std(fold_mae),
        'Mean RMSE': np.mean(fold_rmse),
        'Std RMSE': np.std(fold_rmse),
        'Mean R2': np.mean(fold_r2),
        'Std R2': np.std(fold_r2)
    })

log_results = pd.DataFrame(results)

display(log_results.round(4))
```

```python
import numpy as np
from xgboost import XGBRegressor

from sklearn.model_selection import GroupKFold, RandomizedSearchCV
from sklearn.metrics import make_scorer, mean_absolute_error

# Convert log predictions back to milliseconds
def original_scale_mae(y_true_log, y_pred_log):
    y_true = np.expm1(y_true_log)
    y_pred = np.maximum(np.expm1(y_pred_log), 0)

    return mean_absolute_error(y_true, y_pred)


# Custom scorer: lower original-scale MAE is better
mae_scorer = make_scorer(
    original_scale_mae,
    greater_is_better=False
)

# Base XGBoost model
xgb_tuning_model = XGBRegressor(
    objective='reg:squarederror',
    random_state=42,
    n_jobs=1
)

# Hyperparameter search space
param_grid = {
    'n_estimators': [300, 500, 700],
    'learning_rate': [0.02, 0.03, 0.05, 0.07],
    'max_depth': [3, 4, 5, 6, 7],
    'subsample': [0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.7, 0.8, 0.9, 1.0],
    'min_child_weight': [1, 3, 5, 7],
    'reg_lambda': [1, 3, 5, 10]
}

# Five-fold group validation
group_cv = GroupKFold(n_splits=5)

# Randomized search
random_search = RandomizedSearchCV(
    estimator=xgb_tuning_model,
    param_distributions=param_grid,
    n_iter=15,
    scoring=mae_scorer,
    cv=group_cv,
    random_state=42,
    n_jobs=1,
    verbose=1,
    refit=True
)

# Train using development data only
random_search.fit(
    X_dev,
    y_dev_log,
    groups=groups_dev
)

print("Best Parameters:")
print(random_search.best_params_)

print("\nBest Cross-Validation MAE (milliseconds):")
print(round(-random_search.best_score_, 4))
```

```python
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Original XGBoost configuration
original_xgb = xgb_model

# Newly tuned XGBoost configuration
tuned_xgb = random_search.best_estimator_

models_to_compare = {
    'Original Log XGBoost': original_xgb,
    'Tuned Log XGBoost': tuned_xgb
}

group_cv = GroupKFold(n_splits=5)

comparison_results = []

for name, model in models_to_compare.items():

    mae_scores = []
    rmse_scores = []
    r2_scores = []

    for train_idx, test_idx in group_cv.split(
        X_dev, y_dev_log, groups_dev
    ):

        X_train_fold = X_dev.iloc[train_idx]
        X_test_fold = X_dev.iloc[test_idx]

        y_train_fold = y_dev_log.iloc[train_idx]
        y_test_fold = y_dev.iloc[test_idx]

        y_test_original = y_dev.iloc[test_idx]

        current_model = clone(model)

        current_model.fit(X_train_fold, y_train_fold)

        predictions_log = current_model.predict(X_test_fold)

        predictions = np.maximum(
            np.expm1(predictions_log), 0
        )

        mae_scores.append(
            mean_absolute_error(y_test_original, predictions)
        )

        rmse_scores.append(
            np.sqrt(
                mean_squared_error(y_test_original, predictions)
            )
        )

        r2_scores.append(
            r2_score(y_test_original, predictions)
        )

    comparison_results.append({
        'Model': name,
        'Mean MAE': np.mean(mae_scores),
        'Std MAE': np.std(mae_scores),
        'Mean RMSE': np.mean(rmse_scores),
        'Std RMSE': np.std(rmse_scores),
        'Mean R2': np.mean(r2_scores),
        'Std R2': np.std(r2_scores)
    })

xgb_comparison = pd.DataFrame(comparison_results)

display(xgb_comparison.round(4))
```

```python
# STEP 6: FEATURE PREPROCESSING

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Separate numerical and categorical columns
numerical_features = [
    'cpu_usage',
    'avg_mem',
    'max_mem',
    'avg_gpu_wrk_mem',
    'max_gpu_wrk_mem',
    'read',
    'write',
    'read_count',
    'write_count'
]

categorical_features = [
    'task_name',
    'gpu_name'
]

# Numerical preprocessing
numerical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Combine both pipelines
preprocessor = ColumnTransformer([
    ('num', numerical_pipeline, numerical_features),
    ('cat', categorical_pipeline, categorical_features)
])

# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data
X_test_processed = preprocessor.transform(X_test)

print("PREPROCESSING COMPLETED")
print("-----------------------")
print("Training shape:", X_train_processed.shape)
print("Testing shape:", X_test_processed.shape)
```
